```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n, p = 4, 0.4
x_vals = np.arange(0, n + 1)
pmf_vals = binom.pmf(x_vals, n, p)
cdf_vals = binom.cdf(x_vals, n, p)

print("Tabel Probabilitas (Python):")
for x, pmf, cdf in zip(x_vals, pmf_vals, cdf_vals):
    print(f"P(X={x}) = {pmf:.4f} | F({x}) = {cdf:.4f}")

prob_A = binom.cdf(2, n, p)
print(f"\nPeluang maksimal 2 modul gagal (X <= 2) = {prob_A:.4f}")
```

```python
from scipy.integrate import quad

def pdf(x):
    return 0.5 * x if 0 < x < 2 else 0

total_luas, _ = quad(pdf, 0, 2)
peluang_int, _ = quad(pdf, 0.5, 1.5)

print(f"Total Luas Integrasi PDF = {total_luas:.4f} (Valid = {np.isclose(total_luas, 1.0)})")
print(f"Peluang P(0.5 < X < 1.5) = {peluang_int:.4f}")
```
